```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
import pandas as pd
load_dotenv()
```

```python
df = pd.read_csv('employee_db.csv')
columns = list(df.columns)
```

```python
llm = ChatGroq(model='openai/gpt-oss-120b',temperature=0)
```

```python
question = 'what is name of employee with the highest salary.'
prompt = '''Role:Act as an experienced python developer.
Task:Write a pandas code to find out answer of question.
Question:{}
Context:dataframe name is df and its column names are as follows:{}
Constraints:Consider all data in uppercase.
Output format: Give executable python code only. Strictly dont provide any other information.print only final answer, 
no any import statements. If you dont know answer, give output None'''.format(question,columns)
```

```python
response = llm.invoke(prompt)
code = response.content
print(code)
```

```python
exec(code)
```
